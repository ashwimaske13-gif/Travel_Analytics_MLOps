```python
# basic info of datasets
def dataset_summary(df, dataset_name):
    print("=" * 70)
    print(f"Dataset : {dataset_name}")
    print("=" * 70)

    print("\nShape")
    print(df.shape)

    print("\nColumns")
    print(df.columns.tolist())

    print("\nData Types")
    print(df.dtypes)

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDuplicate Rows")
    print(df.duplicated().sum())
```

```python
dataset_summary(users_df, "Users")
```

```python
dataset_summary(flights_df, "Flights")
```

```python
dataset_summary(hotels_df, "Hotels")
```

```python
# /to check unique values in each column of the datasets
def check_unique_values(df, dataset_name):
    print(f"\n{'='*60}")
    print(f"Unique Values in {dataset_name}")
    print(f"{'='*60}")

    for column in df.columns:
        print(f"\nColumn: {column}")
        print(f"Unique Values: {df[column].nunique()}")
```

```python
check_unique_values(users_df, "Users")
```

```python
check_unique_values(flights_df, "Flights")
```

```python
check_unique_values(hotels_df, "Hotels")
```

```python
# checking missing value percentage in each dataset
def missing_percentage(df):
    missing = pd.DataFrame({
        "Missing Values": df.isnull().sum(),
        "Percentage": (df.isnull().sum() / len(df)) * 100
    })

    return missing.sort_values("Percentage", ascending=False)
```

```python
missing_percentage(users_df)
```

```python
missing_percentage(flights_df)
```

```python
missing_percentage(hotels_df)
```
