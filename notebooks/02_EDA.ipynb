```python
# ==========================================
# Import Required Libraries
# ==========================================

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

plt.style.use("ggplot")
sns.set_theme(style="whitegrid")

print("Libraries Imported Successfully")
```

```python
# Load Users Dataset
users_df = pd.read_csv("../data/raw/users.csv")

# Load Flights Dataset
flights_df = pd.read_csv("../data/raw/flights.csv")

# Load Hotels Dataset
hotels_df = pd.read_csv("../data/raw/hotels.csv")
```

```python
# Gender Distribution
plt.figure(figsize=(6,5))

sns.countplot(
    data=users_df,
    x='gender'
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")

plt.show()
```

Most users belong to one gender category.
This helps us understand class balance for the classification model.

```python
# Age Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    users_df['age'],
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
# Boxplot of Age
plt.figure(figsize=(6,4))

sns.boxplot(
    x=users_df['age']
)

plt.title("Age Distribution with Outliers")

plt.show()
```

```python
# Company Distribution
plt.figure(figsize=(12,6))

users_df['company'].value_counts().head(10).plot(kind='bar')

plt.title("Top 10 Companies")
plt.xlabel("Company")
plt.ylabel("Number of Users")

plt.xticks(rotation=45)

plt.show()
```

```python
# Top Destination Airports
plt.figure(figsize=(12,5))

flights_df['to'].value_counts().head(10).plot(kind='bar')

plt.title("Top Destination Airports")

plt.show()
```

```python
# Total Bill Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    hotels_df['total'],
    bins=25,
    kde=True
)

plt.title("Hotel Total Cost Distribution")

plt.show()
```

```python
# Correlation Matrix (Flights)
plt.figure(figsize=(8,6))

numeric_df = flights_df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix - Flights Dataset")

plt.show()
```

```python
# Correlation Matrix (Hotels)
plt.figure(figsize=(8,6))

numeric_df = hotels_df.select_dtypes(include=np.number)

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='viridis'
)

plt.title("Correlation Matrix - Hotels Dataset")

plt.show()
```
